```
# p-tau217 density colored by hyperlipidemia (matches plot_tau217_kde_by_factor palette)
from scipy.stats import mannwhitneyu

hyperlipidemia_palette = {"NO": "#b4cde3", "YES": "#fbb4ae"}
hue_order = ["NO", "YES"]

plot_df_tau = df_high[["Hyperlipidemia", "p-tau217"]].dropna().copy()
plot_df_tau = plot_df_tau[plot_df_tau["Hyperlipidemia"].isin(hue_order)]

tau_no_hyper = plot_df_tau.loc[plot_df_tau["Hyperlipidemia"] == "NO", "p-tau217"]
tau_yes_hyper = plot_df_tau.loc[plot_df_tau["Hyperlipidemia"] == "YES", "p-tau217"]
wilcoxon_p = mannwhitneyu(
    tau_no_hyper,
    tau_yes_hyper,
    alternative="two-sided",
).pvalue
sig_label = default_significance_label(wilcoxon_p)
print(
    f"Mann-Whitney U (Wilcoxon rank-sum): NO n={len(tau_no_hyper)}, "
    f"YES n={len(tau_yes_hyper)}, p={wilcoxon_p:.4g} ({sig_label})"
)

configure_illustrator_fonts()

fig, ax = plt.subplots(figsize=(10, 5))
sns.kdeplot(
    data=plot_df_tau,
    x="p-tau217",
    hue="Hyperlipidemia",
    hue_order=hue_order,
    palette=hyperlipidemia_palette,
    fill=True,
    common_norm=False,
    alpha=0.5,
    ax=ax,
)
ax.set_xlabel("p-tau217")
ax.set_title(
    f"p-tau217 KDE by hyperlipidemia status, "
    f"N NO = {len(tau_no_hyper)}, N YES = {len(tau_yes_hyper)}"
)
ax.text(
    0.98,
    0.98,
    f"Mann-Whitney p = {wilcoxon_p:.4g} ({sig_label})",
    transform=ax.transAxes,
    ha="right",
    va="top",
    fontsize=10,
)
leg = ax.get_legend()
if leg is not None:
    leg.set_title("Hyperlipidemia")
fig.tight_layout()

save_fig_for_illustrator(fig, f"{SAVE_DIR}/tau217_kde_by_hyperlipidemia.pdf")
plt.show()
```

```
# p-tau217 density by uncontrolled hyperlipidemia (LDL + non-HDL thresholds)
from scipy.stats import mannwhitneyu

ldl_thr = 116
non_hdl_thr = ldl_thr + 30

plot_df_tau = df_high[["Hyperlipidemia", "p-tau217", "LDL", "Cholesterol", "HDL"]].dropna().copy()
plot_df_tau = plot_df_tau[plot_df_tau["Hyperlipidemia"].isin(["NO", "YES"])]
plot_df_tau["non_HDL_high"] = (plot_df_tau["Cholesterol"] - plot_df_tau["HDL"]) > non_hdl_thr
plot_df_tau["Uncontrolled_hyperlipidima"] = (
    (plot_df_tau["Hyperlipidemia"] == "YES") &
    (plot_df_tau["LDL"] > ldl_thr)
    & plot_df_tau["non_HDL_high"]
)
# String labels so hue_order/palette match what seaborn plots
plot_df_tau["hyper_group"] = np.where(
    plot_df_tau["Uncontrolled_hyperlipidima"],
    "Uncontrolled",
    "Controlled",
)
hue_order = ["Controlled", "Uncontrolled"]
hyperlipidemia_palette = {
    "Controlled": "#b4cde3",
    "Uncontrolled": "#fbb4ae",
}

tau_not_uncontrolled = plot_df_tau.loc[
    plot_df_tau["hyper_group"] == "Controlled", "p-tau217"
]
tau_uncontrolled = plot_df_tau.loc[
    plot_df_tau["hyper_group"] == "Uncontrolled", "p-tau217"
]
wilcoxon_p = mannwhitneyu(
    tau_not_uncontrolled,
    tau_uncontrolled,
    alternative="two-sided",
).pvalue
sig_label = default_significance_label(wilcoxon_p)
print(
    f"Mann-Whitney U: Controlled n={len(tau_not_uncontrolled)}, "
    f"Uncontrolled n={len(tau_uncontrolled)}, p={wilcoxon_p:.4g} ({sig_label})"
)

configure_illustrator_fonts()

fig, ax = plt.subplots(figsize=(10, 5))
sns.kdeplot(
    data=plot_df_tau,
    x="p-tau217",
    hue="hyper_group",
    hue_order=hue_order,
    palette=hyperlipidemia_palette,
    fill=True,
    common_norm=False,
    alpha=0.5,
    ax=ax,
)
ax.set_xlabel("p-tau217")
ax.set_title(
    f"p-tau217 KDE by uncontrolled hyperlipidemia "
    f"(LDL>{ldl_thr}, non-HDL>{non_hdl_thr}), "
    f"N controlled = {len(tau_not_uncontrolled)}, "
    f"N uncontrolled = {len(tau_uncontrolled)}"
)
ax.text(
    0.98,
    0.98,
    f"Mann-Whitney p = {wilcoxon_p:.4g} ({sig_label})",
    transform=ax.transAxes,
    ha="right",
    va="top",
    fontsize=10,
)
leg = ax.get_legend()
if leg is not None:
    leg.set_title("Hyperlipidemia control")
fig.tight_layout()

save_fig_for_illustrator(fig, f"{SAVE_DIR}/tau217_kde_by_hyperlipidemia.pdf")
plt.show()
```

```
# Biomarkers to show in one combined grid
features_to_plot = [
    "LDL",
    "HDL",
    "Cholesterol",
    "Triglycerides",
    "Hemoglobin_A1C",
    "CRP",
    "Lymphocytes",
    "SBP",
    "DBP",
]

# Which risk-factor column should be used for each biomarker
feature_group_col = {
    "LDL": "Hyperlipidemia",
    "HDL": "Hyperlipidemia",
    "Cholesterol": "Hyperlipidemia",
    "Triglycerides": "Hyperlipidemia",
    "Hemoglobin_A1C": "Diabetes",
    "CRP": "Recurrent_UTI/_pneumonia/other_infection",
    "Lymphocytes": "Recurrent_UTI/_pneumonia/other_infection",
    "SBP": "Hypertension",
    "DBP": "Hypertension",
}

# Readable x-axis labels for each risk-factor column
group_col_label = {
    "Hyperlipidemia": "Hyperlipidemia",
    "Diabetes": "Diabetes",
    "Recurrent_UTI/_pneumonia/other_infection": "Recurrent infections",
    "Hypertension": "Hypertension",
}

# Make sure feature columns are numeric
for col in features_to_plot:
    df_high_rf[col] = pd.to_numeric(df_high_rf[col], errors="coerce")

# Y-axis labels
y_labels = {
    "LDL": "LDL (mg/dL)",
    "HDL": "HDL (mg/dL)",
    "Cholesterol": "Cholesterol (mg/dL)",
    "Triglycerides": "Triglycerides (mg/dL)",
    "Hemoglobin_A1C": "Hemoglobin A1C (%)",
    "CRP": "CRP (mg/dL)",
    "Lymphocytes": "Lymphocytes",
    "SBP": "Systolic BP (mmHg)",
    "DBP": "Diastolic BP (mmHg)",
}

# Thresholds per biomarker (single value or list of values)
thresholds = {
    "LDL": 100,
    "HDL": 40,
    "Cholesterol": 200,
    "Triglycerides": 150,
    "Hemoglobin_A1C": 5.7,
    "CRP": 1,
    "Lymphocytes": [1, 4.8],
    "SBP": 130,
    "DBP": 80,
}

# Layout
n_feat = len(features_to_plot)
nrows = 3
ncols = int(np.ceil(n_feat / nrows))
fig_width = 5 * ncols
fig_height = 5 * nrows

fig, axes = plt.subplots(nrows=nrows, ncols=ncols, figsize=(fig_width, fig_height), sharey=False)
axes_flat = axes.flatten() if hasattr(axes, "flatten") else np.array([axes])

palette = ["#b4cde3", "#fbb4ae"]

def _ordered_groups(series):
    vals = [str(v) for v in series.dropna().unique()]
    if "NO" in vals and "YES" in vals:
        return ["NO", "YES"]
    return sorted(vals)

for i, feat in enumerate(features_to_plot):
    ax = axes_flat[i]
    group_col = feature_group_col[feat]

    plot_df = df_high_rf[[feat, group_col]].copy()
    plot_df[group_col] = plot_df[group_col].astype(str)
    order = _ordered_groups(plot_df[group_col])

    sns.boxplot(
        data=plot_df,
        x=group_col,
        y=feat,
        order=order,
        palette=palette[: len(order)] if len(order) <= len(palette) else None,
        ax=ax,
    )

    ax.set_title(feat)
    ax.set_xlabel(group_col_label.get(group_col, group_col))
    ax.set_ylabel(y_labels.get(feat, feat))

    th = thresholds.get(feat)
    if th is not None:
        th_values = th if isinstance(th, (list, tuple, np.ndarray)) else [th]
        for k, th_val in enumerate(th_values):
            label = f"Threshold: {th_val}" if k == 0 else None
            ax.axhline(th_val, color="red", linestyle="--", label=label)
        ax.legend()

# Hide extra subplot(s) if grid has spare axes
for j in range(n_feat, len(axes_flat)):
    axes_flat[j].axis("off")

fig.suptitle("Various Biomarkers by Matched Risk-Factor Group", fontsize=18)
fig.tight_layout(rect=[0, 0, 1, 0.96])

# Save and show
plt.savefig(f"{SAVE_DIR}/biomarkers_by_matched_risk_factor_group.pdf")
plt.show()
```

```
# Single plot: horizontal boxplots on a shared unitless scale (min-max per biomarker)
features_to_plot = [
    "Hemoglobin_A1C",
    "Cholesterol",
    "LDL",
    "Cholesterol_HDL_ratio",
    "Triglycerides",
    "CRP",
    "DBP",
    "SBP",
]

feature_titles = {
    "Hemoglobin_A1C": "Hemoglobin A1C",
    "Cholesterol": "Total Cholesterol",
    "LDL": "LDL",
    "Cholesterol_HDL_ratio": "Total Cholesterol/HDL",
    "Triglycerides": "Triglycerides",
    "CRP": "CRP",
    "DBP": "DBP",
    "SBP": "SBP",
}

thresholds = {
    "Hemoglobin_A1C": 5.7,
    "Cholesterol": 200,
    "LDL": 100,
    "Cholesterol_HDL_ratio": 3.5,
    "Triglycerides": 150,
    "CRP": 1,
    "DBP": 80,
    "SBP": 130,
}

for col in features_to_plot:
    df_high_rf[col] = pd.to_numeric(df_high_rf[col], errors="coerce")

def _minmax_scale(series, value):
    """Scale value to [0, 1] using the observed min/max of series."""
    vmin, vmax = series.min(), series.max()
    if pd.isna(vmin) or pd.isna(vmax) or vmax == vmin:
        return np.nan
    return (value - vmin) / (vmax - vmin)

# Long format with per-biomarker min-max normalization (comparable x-axis, no raw units)
rows = []
for feat in features_to_plot:
    vals = df_high_rf[feat].dropna()
    vmin, vmax = vals.min(), vals.max()
    if vmax == vmin:
        scaled = pd.Series(0.5, index=vals.index)
    else:
        scaled = (vals - vmin) / (vmax - vmin)
    for v in scaled:
        rows.append({"feature": feat, "label": feature_titles[feat], "scaled": v})
plot_df = pd.DataFrame(rows)

label_order = [feature_titles[f] for f in features_to_plot]

fig, ax = plt.subplots(figsize=(10, 8))
sns.boxplot(
    data=plot_df,
    y="label",
    x="scaled",
    order=label_order,
    orient="h",
    color=healthy_color,
    width=0.55,
    showfliers=False,
    ax=ax,
)

# Per-biomarker threshold lines (clinical value noted in label; x is unitless)
for i, feat in enumerate(features_to_plot):
    vals = df_high_rf[feat].dropna()
    th_val = thresholds.get(feat)
    if th_val is None or vals.empty:
        continue
    th_scaled = _minmax_scale(vals, th_val)
    if pd.isna(th_scaled):
        continue
    ax.vlines(
        th_scaled,
        i - 0.38,
        i + 0.38,
        colors="red",
        linestyles="--",
        linewidth=1.5,
        zorder=5,
    )
    ax.text(
        th_scaled,
        i + 0.42,
        f"thr={th_val}",
        color="red",
        fontsize=8,
        ha="center",
        va="bottom",
    )

ax.set_xlabel("Relative position within cohort (min–max scaled, unitless)")
ax.set_ylabel("")
ax.set_xticks([])  # raw / scaled numbers are not meaningful across biomarkers
ax.set_title(f"Biomarker distributions (high p-tau217 cohort, N={len(df_high_rf)})\nRed dashed line = clinical threshold per biomarker")
fig.tight_layout()
plt.savefig(f"{SAVE_DIR}/biomarkers_horizontal_combined_thresholds.pdf", bbox_inches="tight")
plt.show()
```
